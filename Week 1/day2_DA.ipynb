# LISTS, TUPLES, SETS AND DICTIONARIES 

```python
#LISTS,TUPLES AND SETS 
#Creating and reading lists 
# A lists holds many values in order, inside square brackets. They are mutable 
# .append, .insert, .extend, .remove, .pop, .reverse, .sort, .index

courses = ['History','Math', 'Physics']
print(courses)
print(courses[-1])
print(courses[0:2])
monthly_sales = [120,150,135,190,210]
print(monthly_sales)

monthly_sales.append(230)
print(sum(monthly_sales))
print(max(monthly_sales),min(monthly_sales))

print('Math' in courses)

for course in courses: 
  print(course)

course_str = ','.join(courses)
print(course_str)

#Tuples 
# They are immmutable, work just like lists except anything function that mutates it 
# They use the usual brackest as opposed to square brackets for sets

tuple_1 = ('History','Math','Chemistry')
tuple_2 = ('Biology','Compsci')

# tuple_1[0] = 'Commerce'  #This displays and error saying tuple does not support item assignment 

#Sets
# unlike tuples and lists sets don't care about order, they are mutable and we use curly brackets
# Used for membership tests and comparisons, also remove duplicates.

cs_courses = {'History','Math','Chemistry','Math'}
print(cs_courses)     #Returns on;y one value for Math

# in , .intersection , .difference , .union 

#Dictionaries 
# Dictionaries stores key-value pairs inside curly brackets, look values up by name instead of position
# .get , .update() , del , .pop , .keys , .values , .items() 

student = {'name':'John','age':25,'courses':['Math','Compsci']}
print(student)

# print(student.get('phone'),'Not Found')

student.update({'name':'Jane','age':28 , 'phone':'555-6739-1425'})
print(student)

del student['age']
print(student)

for key, value in student.items():
 print(key ,':', value)

```

```python
#Simple interest
# A loan of 500000 at 15% per year for 2 years. Store the principal, rate and years in variables. Print the interest and the total to repay.
Principal = 500000
rate = 0.15
Years = 2

Interest = Principal * rate * Years 
Total_Repayment = Principal + Interest 

print(Interest)
print(Total_Repayment)
```

```python
# Percentage Change 
# A product price falls from 1200 to 1050. Print the percentage change using an f-string with one decimal place.
Product_price = 1200
New_price = 1050 

percent_change = (New_price - Product_price)/Product_price * 100
print(f"{percent_change:.1f}""%")
```

```python
#Compound Interest
# 200000 invested at 10% per year for 5 years. Print the final amount formatted with commas and 2 decimals.
Principal = 200000
rate = 0.10
Years = 5 

compound_interest = Principal*(1+rate)**Years
print(f"{compound_interest: ,.2f}")
```

```python
#Type conversion
# The text "2,500" arrives from a messy file. Remove the comma, convert it to an integer, add 500, and print the result.
text = "2,500"
new_text = int(text.replace('2,500','2500'))
print(new_text + 500)
```

```python
#String Cleaning 
#Clean " gtbank transfer " so it has no outer spaces and each word starts with a capital letter. Print the cleaned text and its length.
method = "  gtbank transfer  "             
method = method.strip().title()         #We can also add both fucntions on the same line
#method = method.title()
print(method,'and',len(method))
```

```python
#Slicing 
# For the account number "0123456789", print the last 4 digits, then a masked version with 6 stars followed by those digits.
account_number = "0123456789"
last4 = account_number[-4:]
print(last4)              
print(f"Masked: {'*' * 6}{last4}")
```

```python
#List Statistics 
# Given daily_balances = [52000, 48500, 61000, 59500, 70200], print the average, maximum, minimum, the range and the last 3 balances.
daily_balances = [52000, 48500, 61000, 59500, 70200]
print(max(daily_balances))
print(min(daily_balances))
Range = (max(daily_balances)-min(daily_balances))
print(Range)
Avg = sum(daily_balances)/len(daily_balances)
print(Avg)

print(daily_balances[-3:])
```

```python
#List editing 
# Start with expenses = [3500, 12000, 800, 4500]. Add 6000, sort from largest to smallest, then print the top 2 expenses and the total.
expenses = [3500, 12000, 800, 4500]
expenses.append(6000)
print(expenses)
expenses.sort(reverse=True)
print(expenses)

print(sum(expenses))
print(expenses[:])
```
