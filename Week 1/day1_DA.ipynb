```python
import sys
import pandas as pd
import numpy as np
import matplotlib
print("Python:", sys.version.split()[0])
print("pandas:", pd.__version__)
print("numpy:", np.__version__)
print("matplotlib:", matplotlib.__version__)
```

```python
import matplotlib.pyplot as plt
data = {"month": ["Jan", "Feb", "Mar", "Apr"],
        "transactions": [120, 150, 135, 190]}
df = pd.DataFrame(data)
print(df)
plt.bar(df["month"], df["transactions"])
plt.title("Monthly transactions (sample data)")
plt.ylabel("Count")
plt.xlabel("months")
plt.show()
```
